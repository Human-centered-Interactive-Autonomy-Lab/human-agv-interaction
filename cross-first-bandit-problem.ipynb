```python
# Get the total number of null values in each column
null_counts = data.isnull().sum()
null_counts
```

```python
# List of columns to check for null values
columns_to_check = ['Age', 'Gender', 'Trust2', 'MWL', 'Assessment']

# Function to get PIDs with null values for a given column
def get_null_pids(dataframe, column):
    return dataframe[dataframe[column].isnull()]['PID'].tolist()

# Dictionary to store results
null_results = {}

for column in columns_to_check:
    if column in data.columns:
        null_pids = get_null_pids(data, column)
        null_count = len(null_pids)
        null_results[column] = {
            'Null Count': null_count,
            'PIDs with Null Values': null_pids
        }
    else:
        print(f"Warning: Column '{column}' not found in the dataset.")

# Print results
for column, result in null_results.items():
    print(f"\n{column}:")
    print(f"Null Count: {result['Null Count']}")
    print("PIDs with Null Values:")
    print(result['PIDs with Null Values'])
```
